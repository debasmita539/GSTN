Detect How many OUTLIERS PREESENT ON THE DATASET(ISOLATION FOREST)#Original_Outlier.csv'

```python

import pandas as pd
from sklearn.ensemble import IsolationForest

# Load the dataset
data = pd.read_csv(r'D:\GST\data2.csv')

# List of columns to check for outliers (you can use all columns if needed)
columns_to_check = [
    'Column0', 'Column1', 'Column2', 'Column3','Column4', 'Column5', 'Column6', 
    'Column7', 'Column8', 'Column9', 'Column10', 'Column11', 'Column12', 
    'Column13', 'Column14', 'Column15', 'Column16', 'Column17', 
    'Column18', 'Column19', 'Column20', 'Column21'
]

# Subset the data to include only the columns to check for outliers
subset_data = data[columns_to_check]

# Initialize Isolation Forest
iso_forest = IsolationForest(contamination=0.01, random_state=42)  # Adjust contamination parameter if needed

# Fit the model
iso_forest.fit(subset_data)

# Predict the outliers (returns -1 for outliers and 1 for inliers)
outliers = iso_forest.predict(subset_data)

# Add the outlier flag to the original dataset
data['Outlier_Flag'] = outliers

# Dictionary to store the number of outliers for each column
outliers_per_column = {}

# Count the number of outliers in each column
for column in columns_to_check:
    outliers_per_column[column] = ((data['Outlier_Flag'] == -1) & (data[column].notna())).sum()

# Print the number of outliers detected in each column
print("Number of outliers detected in each column:")
for column, num_outliers in outliers_per_column.items():
    print(f"{column}: {num_outliers} outliers")


# Save the dataset with outlier flags
data.to_csv(r'Original_Outlier.csv', index=False)

print("Dataset with outlier flags saved to 'Original_Outlier.csv")

```

Detect outlier using IQR 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset
data2 = pd.read_csv(r'D:\GST\data2.csv')

# List of columns to check for outliers
columns_to_check = ['Column0','Column1','Column2', 'Column3','Column4', 'Column5', 'Column6', 'Column7', 'Column8', 'Column9', 'Column10', 'Column11','Column12','Column13','Column14', 'Column15', 'Column16', 'Column17', 'Column18', 'Column19', 'Column20', 'Column21']

# Initialize a dictionary to store the number of outliers in each column
outliers_count = {}

# Loop through each column
for column in columns_to_check:
    # Calculate IQR and bounds for outlier detection
    Q1 = data2[column].quantile(0.25)
    Q3 = data2[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Create a mask to identify rows with outliers
    mask = (data2[column] < lower_bound) | (data2[column] > upper_bound)

    # Count the number of outliers
    outliers_count[column] = mask.sum()

    # Replace outlier values with the median of the column
    median_value = data2[column].median()
    data2.loc[mask, column] = median_value

# Save the cleaned dataset
#data2.to_csv(r'Hey_data_clean.csv', index=False)

# Display the number of outliers detected in each column
print("Number of outliers detected in each column:")
for column, count in outliers_count.items():
    print(f"{column}: {count} outliers")

# Optional: You can also display the total number of outliers
total_outliers = sum(outliers_count.values())
print(f"Total number of outliers detected: {total_outliers}")



```

Detect outlier using a hybrid approach of(log_trasformation + IQR) Method


```python
import pandas as pd
import numpy as np

# Load the dataset
data = pd.read_csv(r'D:\GST\data2.csv')

# List of columns to check for outliers (without duplicates)
columns_to_check = [  'Column0','Column1','Column2', 'Column3', 'Column5', 'Column6', 'Column7', 'Column8', 'Column9', 'Column10','Column11', 'Column12', 'Column13',  'Column14', 'Column15', 'Column16', 'Column17', 'Column18', 'Column19', 'Column20', 'Column21'
]

# Function to detect outliers using log transformation and IQR method
def detect_outliers_log_iqr(df, column):
    # Apply log transformation (adding 1 to avoid log(0))
    df_log = np.log1p(df[column])
    
    # Calculate Q1 (25th percentile) and Q3 (75th percentile)
    Q1 = df_log.quantile(0.25)
    Q3 = df_log.quantile(0.75)
    IQR = Q3 - Q1  # Interquartile Range (IQR)
    
    # Calculate lower and upper bounds for detecting outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Detect outliers in the original data (not transformed)
    outliers = df[(df[column] < np.exp(lower_bound) - 1) | (df[column] > np.exp(upper_bound) - 1)]
    return outliers

# Dictionary to store the number of outliers for each column
outliers_dict = {}

# Iterate over each column and detect outliers using log transformation and IQR
for column in columns_to_check:
    outliers = detect_outliers_log_iqr(data, column)
    num_outliers = len(outliers)
    outliers_dict[column] = num_outliers

#  number of outliers detected in each column
print("Number of outliers detected in each column (using log transformation and IQR):")
for column, num_outliers in outliers_dict.items():
    print(f"{column}: {num_outliers} outliers")

total_outliers = sum(outliers_dict.values())
print(f"Total number of outliers detected: {total_outliers}")

# Save the results to a CSV file
outliers_summary = pd.DataFrame(list(outliers_dict.items()), columns=['Column', 'Num_Outliers'])
outliers_summary.to_csv('Outliers_Summary_Log_IQR.csv', index=False)

print("Outliers summary saved to 'Outliers_Summary_Log_IQR.csv'.")
```

```python
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.preprocessing import MinMaxScaler  

# Load data
x_train = pd.read_csv(r'D:\GST\selected_features_data.csv') 
y_train = pd.read_csv(r'D:\GST\Train_60\Train_60\Train_60\y_train_target_data.csv')  
x_test = pd.read_csv(r'D:\GST\Test_20\Test_20\Test_20\X_Test_Data_Input.csv')  
y_test = pd.read_csv(r'D:\GST\Test_20\Test_20\Test_20\Y_Test_Data_Target.csv')  

# Drop the 'ID' column if needed
column_to_exclude = 'ID'
x_train = x_train.drop(columns=[column_to_exclude], errors='ignore')
x_test = x_test.drop(columns=[column_to_exclude], errors='ignore')

# Ensure x_train and x_test have the same columns
x_test = x_test[x_train.columns]

# Fill NA values in x_test with zeros
x_test.fillna(0, inplace=True)

# Ensure y_train and y_test are 1D arrays and check their shapes
y_train_values = y_train.iloc[:, 1].values.ravel() 
y_test_values = y_test.iloc[:, 1].values.ravel() 

# Apply SMOTE to balance the training data
smote = SMOTE(random_state=42)
x_train_resampled, y_train_resampled = smote.fit_resample(x_train, y_train_values)

# Normalize the training and test data using Min-Max Scaling AFTER SMOTE
scaler = MinMaxScaler()

# Fit the scaler on the training data (resampled)
x_train_normalized = scaler.fit_transform(x_train_resampled)

# Transform the test data based on the scaler fit on training data
x_test_normalized = scaler.transform(x_test)

# Train the Random Forest model on the resampled and normalized data
model = RandomForestClassifier(random_state=42)
model.fit(x_train_normalized, y_train_resampled)

# Evaluate the model on the normalized test data
y_pred = model.predict(x_test_normalized)
print(classification_report(y_test_values, y_pred))
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, roc_curve, auc
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Step 1: Load the normalized data
x = pd.read_csv(r'D:\GST\x_train_normalized.csv')
y = pd.read_csv(r'D:\GST\y_train_resampled.csv')['target'].values  # Load balanced target values

# Step 2: Split the dataset into training and test sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Step 3: Define the ANN model
model = Sequential()
model.add(Dense(64, input_dim=x_train.shape[1], activation='relu'))  # Input layer
model.add(Dense(32, activation='relu'))  # Hidden layer 1
model.add(Dense(16, activation='relu'))  # Hidden layer 2
model.add(Dense(1, activation='sigmoid'))  # Output layer for binary classification

# Step 4: Compile the model
model.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

# Step 5: Train the model
history = model.fit(x_train, y_train, epochs=50, batch_size=32, validation_data=(x_test, y_test), verbose=1)

# Step 6: Make predictions
y_pred_proba = model.predict(x_test)
y_pred = (y_pred_proba > 0.5).astype('int32')

# Step 7: Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Step 8: Print metrics
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"AUC-ROC: {auc_roc:.4f}")

# Step 9: Print the classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 10: Plot AUC-ROC curve
fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc="lower right")
plt.show()
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, classification_report, confusion_matrix)
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the normalized data
x_train_normalized = pd.read_csv(r'D:\GST\x_train_normalized.csv')
y_train_resampled = pd.read_csv(r'D:\GST\y_train_resampled.csv')['target'].values  # Load balanced target values
x_test_normalized = pd.read_csv(r'D:\GST\x_test_normalized.csv')

# Step 2: Load and clean the y_test data
y_test = pd.read_csv(r'D:\GST\Test_20\Test_20\Test_20\Y_Test_Data_Target.csv')
y_test = y_test.iloc[:, 1].values.ravel()  # Drop the ID column and convert to 1D array

# Step 3: Check for NaN values and handle them
x_train_normalized.fillna(x_train_normalized.mean(), inplace=True)
x_test_normalized.fillna(x_test_normalized.mean(), inplace=True)

# Step 4: Scale features
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train_normalized)
x_test_scaled = scaler.transform(x_test_normalized)

# Step 5: Define the Random Forest Classifier
model = RandomForestClassifier(random_state=42, class_weight='balanced')

# Step 6: Define hyperparameters and their values to search
param_grid = {
    'n_estimators': [50, 100, 200],  # Number of trees in the forest
    'max_depth': [None, 10, 20, 30],  # Maximum depth of the tree
    'min_samples_split': [2, 5, 10],  # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4],     # Minimum number of samples required to be at a leaf node
}

# Step 7: Set up GridSearchCV
grid_search = GridSearchCV(estimator=model, param_grid=param_grid,
                           scoring='roc_auc', cv=3, n_jobs=-1, verbose=2)

# Step 8: Fit GridSearchCV
grid_search.fit(x_train_scaled, y_train_resampled)

# Step 9: Get the best parameters and estimator
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

print("Best Hyperparameters:", best_params)

# Step 10: Make predictions with the best model
y_pred = best_model.predict(x_test_scaled)
y_pred_proba = best_model.predict_proba(x_test_scaled)[:, 1]

# Step 11: Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='binary')
recall = recall_score(y_test, y_pred, average='binary')
f1 = f1_score(y_test, y_pred, average='binary')
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Step 12: Print metrics
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"AUC-ROC: {auc_roc:.4f}")

# Step 13: Print the classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 14: Visualization of predictions
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_pred_proba, y=y_test, alpha=0.6, edgecolor='k')

# Add labels and title
plt.title('Random Forest Classifier Predictions')
plt.xlabel('Predicted Probability of Positive Class')
plt.ylabel('True Labels (0 or 1)')
plt.axhline(y=0.5, color='r', linestyle='--')  # Threshold line for classifying as 1
plt.xlim(0, 1)
plt.ylim(-0.1, 1.1)

# Show the plot
plt.grid()
plt.show()

# Step 15: Confusion Matrix and Heatmap
cm = confusion_matrix(y_test, y_pred)

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Negative (0)', 'Positive (1)'],
            yticklabels=['Negative (0)', 'Positive (1)'])

# Add labels and title
plt.title('Confusion Matrix Heatmap for Random Forest Classifier')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')

# Show the plot
plt.show()
```
